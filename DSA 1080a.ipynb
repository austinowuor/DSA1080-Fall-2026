```python
#if statement
a=33
b=200
if b > a:
  print("b is greater than a")
```

```python
#elif statement
a=33
b=33
if b > a:
  print("b is greater than a")
elif a==b:
  print("a and b are equal")  
```

```python
#else
a=33
b=33
if b > a:
  print("b is greater than a")
elif a==b:
  print("a and b are equal")
else:
  print("a is greater than b")
```

```python
#short and if
a=200
b=33
if a > b: print("a is greater than b")
```

```python
#multiple else statements
a=200
b=33
print("A") if a > b else print("=") if a==b else print("B")
```

```python
#and statement
a=200
b=33
c=500

if a > b and c > a:
  print("Both conditions are True")
```

```python
#or statement
a=200
b=33
c=500

if a > b or c > a:
  print("At least one of the conditions is True")   
```

```python
#nested if statements
x=20
if x > 10:
  print("Above ten,")
  if x > 20:
    print("and also above 20!")
  else:
    print("but not above 20.")
```

```python
#while loop
i=1
while i < 10:
  print(i)
  i += 1
```

```python
#while loop with break statement
i=1
while i < 10:
    print(i)
    if i == 5:
        break
    i += 1


```

```python
#for loop
fruits=["apple", "banana", "cherry"]
for x in fruits:
  print(x)
```

```python
#looping through words
for x in "banana":
  print(x)
```

```python
#break statement within for loop
fruits=["apple", "banana", "cherry"]
for x in fruits:
  print(x)
  if x == "banana":
    break
```

```python
#break statement within for loop
fruits=["apple", "banana", "cherry"]
for x in fruits:
    if x == "banana":
        break
    print(x)
```

```python
#continue statement within for loop
fruits=["apple", "banana", "cherry"]
for x in fruits:
    if x == "banana":
        continue
    print(x)
```

```python
#range function
for x in range(6):
  print(x)
```

```python
#nested for loops
adjectives=["red", "big", "tasty"]
fruits=["apple", "banana", "cherry"]
for x in adjectives:
  for y in fruits:
    print(x, y)
```

```python
#load class csv file
import pandas as pd
df = pd.read_csv('DSA1080_Pandas_Student_Data.csv')

print(df.head())
print(df.shape)

#inspect the data
print(df.info())
print(df.describe())
print(df.isna().sum())
```

```python
#selecting columns
print(df['Name'])
print(df[['Name', 'Score']])
print(df['Score'].mean())
```

```python
#selecting rows with loc
print(df.loc[0])
print(df.loc[0, "Name"])
print(df.loc[0:4,["Name","Score"]])
```

```python
#selecting rows with iloc
print(df.iloc[0:5])
```

```python
#filter table with a condition
high=df[df['Score'] >= 70]
print(high[['Name', 'Score']])

city=df[df['City'] == 'Nairobi']
print(city.head())
```

```python
#Understanding missing values
df=pd.read_csv('DSA1080_Pandas_Student_Data.csv')
print(df.isna().sum())
print(df[df['Score'].isna()])
```
